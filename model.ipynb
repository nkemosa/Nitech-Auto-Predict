```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("car.csv")
df
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
le
```

```python
df["Brand"] = le.fit_transform(df["Brand"])
df["Fuel_Type"] = le.fit_transform(df["Fuel_Type"])
df["Transmission"] = le.fit_transform(df["Transmission"])
df["Model"] = le.fit_transform(df["Model"])
df
```

```python
x = df.drop(["Car_ID", "Price"], axis=1)
x
```

### LINEAR REGRESISON

```python
from sklearn.linear_model import LinearRegression
```

```python
TLRM = LinearRegression()
```

```python
TLRM
```

```python
x = df[[
        "Brand",
        "Model",
        "Year",
        "Engine_Size",
        "Mileage",
        "Fuel_Type",
        "Transmission",
        "Owner_Count"
    ]]
x
```

```python
y = df["Price"]
y
```

### SPLITING

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
TLRM.fit(x_train.values, y_train.values)
TLRM
```

```python
PLRM = TLRM.predict(x_test.values)
PLRM
```

### Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
TDTM = DecisionTreeRegressor(random_state=42)
```

```python
TDTM.fit(x_train, y_train)
```

```python
PDTM = TDTM.predict(x_test)
```

```python
PDTM
```

### RANDOM FOREST

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
TRFM = RandomForestRegressor(n_estimators=200,random_state=42)
```

```python
TRFM.fit(x_train, y_train)
```

```python
PRFM = TRFM.predict(x_test)
```

```python
PRFM
```

### MODAL EVALUATION

```python
# Mean Absolute Error (MAE)
from sklearn.metrics import mean_absolute_error
```

```python
mean_absolute_error(y_test, PLRM)
```

```python
# Mean Squared Error (MSE)
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y_test, PLRM)
```

```python
# Root Mean Squared Error (RMSE)
np.sqrt(mean_squared_error(y_test, PLRM))
```

```python
from sklearn.metrics import r2_score
```

```python
# R² Score (Coefficient of Determination) → closest to "accuracy"
r2_score(y_test, PLRM)
```
